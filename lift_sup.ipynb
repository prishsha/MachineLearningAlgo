```python
data = [["Milk", "Bread"],
        ["Milk", "Butter"],
        ["Bread", "Butter"],
        ["Milk", "Bread", "Butter"]]
rule = [["Milk", "Bread"], ["Butter"]]
```

```python
def count_freq(itemset):
    count=0
    for row in data:
        if all(item in row for item in itemset):
            count+=1
    return count
```

```python
premise = rule[0]
count_pre = count_freq(premise)
print(count_pre)
concl = rule[1]
count_conc = count_freq(concl)
print(count_conc)
comb = list(set(rule[0] + rule[1]))
count_comb = count_freq(comb)
print(count_comb)
```

```python
tot = len(data)
sup_pre = count_pre/tot
print("Support for premise:", sup_pre)
sup_conc = count_conc/tot
print("Support for conclusion:", sup_conc)
sup_both = count_comb/tot
print("Support for both:", sup_both)
lift = sup_both/(sup_pre*sup_conc)
print(lift)
```

```python
conf = sup_both/sup_pre
print("Confidence:", conf)
```
